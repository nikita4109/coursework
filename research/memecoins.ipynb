```python
import pandas as pd
```

```python
df = pd.read_csv('../data/pair_created_2.csv')

pools = set(df['pool'].unique())
file_path = '../data_collector/output_2/transfers.csv'
output_path = '../data/lp_transfers_2.csv'
chunk_size = 100000 

def filter_chunk(chunk, pools):
    return chunk[chunk.iloc[:, 1].isin(pools)]

with open(output_path, 'w') as output_file:
    for chunk in pd.read_csv(file_path, chunksize=chunk_size, header=None):
        filtered_chunk = filter_chunk(chunk, pools)
        
        if not filtered_chunk.empty:
            filtered_chunk.to_csv(output_file, index=False, header=False, mode='a')
```
